```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data
from skimage.feature import canny
from scipy import ndimage as ndi
```

### 02 - Segmentação baseada em bordas (Edge based using Canny)

```python
image = data.coins()

#sigma controla o nível de suavização inicial para reduzir o ruído
#sigma maior detecta bordas em escalas maiores(less details)
edges = canny(image, sigma=5)

fig, axes = plt.subplots(1,2, figsize=(8,5))
ax = axes.ravel()
ax[0].imshow(image, cmap='gray')
ax[0].set_title('Original', fontsize=16)
ax[0].set_axis_off()
ax[1].imshow(edges, cmap='gray')
ax[1].set_title('Bordas Detectadas (Canny Edges)', fontsize=16)
ax[1].set_axis_off()
plt.tight_layout()
plt.show()
```

```python
#Preencher os contornos das moedas

filled_mask_from_edges = ndi.binary_fill_holes(edges)

fig, ax = plt.subplots(1,3, figsize=(20,5))

ax[0].imshow(image, cmap='gray')
ax[0].set_title('Original', fontsize=16)
ax[1].imshow(edges, cmap='gray')
ax[1].set_title('Canny Edges', fontsize=16)
ax[2].imshow(filled_mask_from_edges, cmap='gray')
ax[2].set_title('Segmentação final', fontsize=16)


for a in ax:
    a.set_axis_off()
```

```python
#Imagens com valores de sigma variados para visualizar a diferença.

i = 1

fig, axes = plt.subplots(1,3, figsize=(20,8))

ax = axes.ravel()

for a in ax:
    a.set_axis_off()
    edges = canny(image, sigma=i)
    a.imshow(edges, cmap='gray')
    a.set_title('Sigma = {}'.format(i), fontsize=16)
    i += 4

```
